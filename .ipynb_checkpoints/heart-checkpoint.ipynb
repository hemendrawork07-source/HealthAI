# Heart Disease Prediction

This notebook follows the same general ML-project workflow as the uploaded `ML_project.ipynb`: import libraries, load data, inspect the dataset, prepare features, split the data, train a Random Forest classifier, evaluate it, and save the trained model.

**Model:** Random Forest Classifier  
**Target:** `target`  
**Model output:** `heart_model.pkl`

## 1. Import Libraries

```python
import numpy as np
import pandas as pd
from sklearn import ensemble
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import joblib
```

## 2. Load Heart Disease Dataset

```python
df = pd.read_csv("../data/heart.csv")

df.head()
```

## 3. Dataset Shape

```python
df.shape
```

## 4. Check Missing Values

```python
df.isnull().sum()
```

## 5. Check Duplicate Rows

```python
df.duplicated().sum()
```

## 6. Identify Categorical and Continuous Columns

```python
categorical_val = []
continous_val = []

for column in df.columns:
    if len(df[column].unique()) <= 10:
        categorical_val.append(column)
    else:
        continous_val.append(column)

print("Categorical columns:")
print(categorical_val)

print("\nContinuous columns:")
print(continous_val)
```

## 7. One-Hot Encoding

```python
categorical_val.remove('target')

dataset = pd.get_dummies(df, columns=categorical_val)

dataset.head()
```

## 8. Select Features and Target

The supplied training code uses these seven input columns:

`cp`, `trestbps`, `chol`, `fbs`, `restecg`, `thalach`, `exang`

The target is `target`.

```python
cols = [
    'cp',
    'trestbps',
    'chol',
    'fbs',
    'restecg',
    'thalach',
    'exang'
]

X = df[cols]
y = dataset.target

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

## 9. Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

print(
    'Shape training set: X:{}, y:{}'.format(
        X_train.shape,
        y_train.shape
    )
)

print(
    'Shape test set: X:{}, y:{}'.format(
        X_test.shape,
        y_test.shape
    )
)
```

## 10. Train Random Forest Classifier

```python
model = ensemble.RandomForestClassifier()

model.fit(X_train, y_train)
```

## 11. Make Predictions

```python
y_pred = model.predict(X_test)

y_pred[:10]
```

## 12. Accuracy

```python
print(
    'Accuracy : {}'.format(
        accuracy_score(y_test, y_pred)
    )
)
```

## 13. Classification Report

```python
clf_report = classification_report(
    y_test,
    y_pred
)

print('Classification report')
print("---------------------")
print(clf_report)
print("_____________________")
```

## 14. Confusion Matrix

```python
cm = confusion_matrix(
    y_test,
    y_pred
)

print("Confusion Matrix:")
print(cm)
```

## 15. Save the Model

```python
joblib.dump(
    model,
    "heart_model.pkl"
)

print("Model saved as heart_model.pkl")
```

## 16. Load the Saved Model and Test One Prediction

The saved model expects the same seven features and the same order used during training.

```python
loaded_model = joblib.load("heart_model.pkl")

sample = X_test.iloc[[0]]

prediction = loaded_model.predict(sample)

print("Input:")
print(sample)

print("\nPrediction:")
print(prediction[0])
```
